```python
import matplotlib.pyplot as plt 
```

```python
years = [1950,1956,1966,1997,2011,2016,2022]
```

```python
events = [ "Turing Test", "AI term introduced","ELIZA" , "Deep Blue" ," Siri","AlphaGo","ChatGPT" ]
```

```python
plt.plot(years,[1,1,1,1,1,1,1],'o-')
for i in range(len(years)):
    plt.text(years[i],1.02,events[i],rotation=45,ha='left')
plt.title("History of Artificial Intelligence")
plt.xlabel("year")
plt.yticks([])
plt.show()
```
