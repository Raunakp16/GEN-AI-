```python
from textblob import TextBlob
```

```python
# Take input from the user
text = input("Enter a sentence to analyze your mood: ")
```

```python
# Create a TextBlob object
blob = TextBlob(text)
```

```python
# Calculate sentiment polarity
polarity = blob.sentiment.polarity
```

```python
# Predict sentiment
if polarity > 0:
    print("Mood: Positive 😊")
elif polarity < 0:
    print("Mood: Negative 😔")
else:
    print("Mood: Neutral 😐")

print("Sentiment Polarity:", polarity)
```
